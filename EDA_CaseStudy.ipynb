```python
import matplotlib.pyplot as plt
```

```python
# distribution of age 
plt.title("Age distribution")
plt.xlabel("Age")
plt.ylabel("Count")
plt.hist(data['Age'],bins=15,color='pink')
plt.show()
```

```python
# distribution of age 
counts,edges,bars=plt.hist(data['Age'],bins=15,color='red')
print(counts)
print(bars)
plt.title("Age distribution")
plt.xlabel("Age")
plt.ylabel("Count")
plt.bar_label(bars)
plt.hist(data['Age'],bins=15,color='pink')
plt.show()
```

```python
index = data['Survived'].value_counts().keys().tolist()
print(index)
labels= data['Survived'].value_counts().values.tolist()
print(labels)
```

```python
#pie chart 
plt.pie(x= [577,314],
        labels= ['M','F'],
        colors=['orange','pink'],
        autopct="%.1f%%",
        shadow=True)
plt.title("Gender")
plt.legend()
plt.show()
```

```python
# pairwise relationship -> scatterplot()
plt.scatter(x=data['Age'],
            y=data['Fare'])
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare Distribution")
plt.show()
```

```python
import plotly as py
import plotly.graph_objects as go # for interactive graphs 
```

```python
from plotly.offline import init_notebook_mode,iplot
init_notebook_mode(connected=True)
```

```python
#lets visualize histogram 
trace = go.Histogram(x = data['Age'],
                    nbinsx=20,
                    marker={'color':'red'})
#layout define for the graph
layout = go.Layout(title="Age Distribution")
# we combine the trace with the layout to display the figure
fig = go.Figure(data=trace,layout = layout)
fig
```

```python
print(index)
print(labels)
```

```python
#Now we will display for piechart
trace= go.Pie(labels=['Dead','Alive'],
              values=labels
             )
layout= go.Layout(title = 'Survived Distribution')
fig= go.Figure(data=trace,layout=layout)
fig.show()
```

```python
#Box plot and Scatter plot 
trace = go.Box(x= data['Pclass'],
               y=data['Age'],
               marker={'color':'red'})
layout= go.Layout(title = "Age vs Pclass Distribution")
fig = go.Figure(data=trace, layout=layout)
fig.show()
```

```python
#Scatter plot representation --> Age vs Fare
trace = go.Scatter(x=data['Age'], y=data['Fare'],
                   mode='markers',
                   marker={'color':'violet'})
layout= go.Layout(
    title="Age vs Fare Distribution")
fig=go.Figure(data=trace,layout=layout)
fig.show()
```

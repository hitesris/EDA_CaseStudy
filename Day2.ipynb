Feature Scaling:

It is a technique to standardize the independent features present in the data in fixed range
-> min max normalization
-> Standard Scaler
-> Robust Scaler

min-max normalization_formula:
x(scaled)=x(i)-x(min)/x(max)-x(min)

```python
#import libraries

import pandas as pd
import numpy as np
import sklearn
```

```python
data = pd.DataFrame({'Price_USA':[80,50,140,25],
                     'Price_INR':[5000,7000,12000,3000]})
data
```

```python
#we will perform mathematical wise

data['Price_USA_minmax']=(data['Price_USA']-data['Price_USA'].min())/(data['Price_USA'].max()-data['Price_USA'].min())
data
```

```python
data['Price_INR_minmax']=(data['Price_INR']-data['Price_INR'].min())/(data['Price_INR'].max()-data['Price_INR'].min())
data
```

```python
from sklearn.preprocessing import MinMaxScaler,StandardScaler
```

```python
min_max=MinMaxScaler()
```

```python
#we need to use fit_transform()

data['Price_USA_minsk']=min_max.fit_transform(data[['Price_USA']])
data
```

```python

data['Price_INR_minsk']=min_max.fit_transform(data[['Price_INR']])
data
```

```python
#standard scaler  z=x(i)-mean/std

data['Price_USA_std']=(data['Price_USA']-data['Price_USA'].mean())/(data['Price_USA'].std())
data
```

```python

data['Price_INR_std']=(data['Price_INR']-data['Price_INR'].mean())/(data['Price_INR'].std())
data
```

```python
scaler=StandardScaler()
```

```python
data['Price_USA_scaler']=scaler.fit_transform(data[['Price_USA']])
data
```

```python
data['Price_INR_scaler']=scaler.fit_transform(data[['Price_INR']])
data
```

```python
##step-1: importing the libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from ucimlrepo import fetch_ucirepo
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.preprocessing import OneHotEncoder,PolynomialFeatures
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
sns.set_style(style='whitegrid')   #background style of graphs
```

```python
##step-2: loading (autompg dataset) and cleaning the data

data=fetch_ucirepo(id=9)
data

x=data.data.features.copy()   #it returns all numerical features
y=data.data.targets['mpg']
y
```

```python
x.describe()

x.info()
```

```python
x['origin'].unique()
x['horsepower'].unique()
```

```python
#first we will transform the horsepower and origin columns
x['horsepower']=pd.to_numeric(x['horsepower'],errors='coerce')
#x['horsepower'].info()
x['origin']=x['origin'].astype(str)
```

```python
x.info()
#All numeric columns into a single collections
numerical_features=['cylinders','displacement','horsepower','weight','acceleration','model_year']
categorical_features=['origin']
x=x[numerical_features+categorical_features]
x.info()
x.isnull().sum()
x.shape
```

```python
# step 3: quick explanatory data analysis 
df= x.copy()
df['mpg']=y
```

```python
df
```

```python
#df
# we are gonna use sumplots() concept to divide the plot into multiple plots 
fig,axes= plt.subplots(1,3,figsize=(16,4))
#plot graphs as per axes 
sns.histplot(df['mpg'],kde=True,ax=axes[0],color='black')
#in similar way we display other graphs in rmng plots 
axes[0].set_title('MPG Distribution')
sns.scatterplot(x=df['weight'],y=df['mpg'],ax=axes[1])
axes[1].set_title("Weight vs MPG Distribution")
sns.scatterplot(x=df['horsepower'],y=df['mpg'],ax=axes[2],color='yellow')
axes[2].set_title("HorsePower vs MPG Distribution")
```

```python
# correlation between our variables
plt.figure(figsize=(16,4))
sns.heatmap(df.select_dtypes(include=np.number).corr(),
           annot = True,
            cmap='coolwarm',
            center=0
           )
plt.title("Correlation matrix")
plt.tight_layout()
plt.show()
```

```python
# step 4: Dividing the data into train tezt split 
#print(x)
#print(y)
#Split rule--> 70-30,80-20,75-25
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
print('\nTrain Test Split')
print("Training Data",len(x_train))
print("Testing Data",len(x_test))
#print(x_train)
```

```python
#step 5: creating a baseline model 
def evaluate(model_name,actual,predicted):
    """Returns MAE,RMSE,R2-SCORE"""
    mse= mean_squared_error(actual,predicted)
    mae= mean_absolute_error(actual,predicted)
    return {'Model_Name':model_name,
    'MAE':mae,
    'MSE':mse,
    'RMSE':np.sqrt(mse),
    'RS-Score':r2_score(actual,predicted)}
results=[]
```

```python
# step 6: Baseline model --> It predicts the training set for every values 
baseline_predictions=np.full(len(y_test),y_train.mean())
#baseline_predictions
# in above case we are keeping a baseline prediction 
results.append(evaluate("Mean baseline",y_test,baseline_predictions))
results
```

```python
# step 7 : we apply linear regression
simple_model=LinearRegression()
# we fit the data to the model
simple_model.fit(x_train[['weight']],y_train)
simple_model_predictions=simple_model.predict(x_test[['weight']])
#print(simple_model_predictions)
results.append(evaluate("Simple Linear Regression:Weight",
                            y_test,
                            simple_model_predictions))
print(len(results))
```

```python
# we will check the intercept value 
print(f'Simple Linear Regression Equation')
print(f"Predicted MPG = {simple_model.intercept_}+{simple_model.coef_[0]}*weight")
```

```python
#Step7: We apply Linear regression for cylinders
simple_model = LinearRegression()
#we fit the data the model
simple_model.fit(x_train[['cylinders']],y_train)
simple_model_Predictions = simple_model.predict(x_test[['cylinders']])
#print(simple_model_predictions)
results.append(evaluate("Simple Linear Regression:cylinders",
                        y_test,
                        simple_model_Predictions))
print(len(results))
```

```python
print(f'Simple Linear Regression Equation')
print(f"Predicted MPG = {simple_model.intercept_}+({simple_model.coef_[0]})*cylinders")
```

```python
#Step7: We apply Linear regression for displacement
simple_model = LinearRegression()
#we fit the data the model
simple_model.fit(x_train[['displacement']],y_train)
simple_model_Predictions = simple_model.predict(x_test[['displacement']])
#print(simple_model_predictions)
results.append(evaluate("Simple Linear Regression:displacement",
                        y_test,
                        simple_model_Predictions))
print(len(results))
```

```python
print(f'Simple Linear Regression Equation')
print(f"Predicted MPG = {simple_model.intercept_}+({simple_model.coef_[0]})*displacement")
```

```python
# Find the mean horsepower from training data
horsepower_mean = x_train['horsepower'].mean()

print("Mean horsepower:", horsepower_mean)
```

```python
# Fill missing horsepower values with the training mean
x_train['horsepower'] = x_train['horsepower'].fillna(horsepower_mean)
x_test['horsepower'] = x_test['horsepower'].fillna(horsepower_mean)
```

```python
print("Missing values in x_train:")
print(x_train['horsepower'].isnull().sum())

print("Missing values in x_test:")
print(x_test['horsepower'].isnull().sum())
```

```python
#Step7: We apply Linear regression for horse power 
df['horsepower'] = df['horsepower'].fillna(df['horsepower'].mean())
simple_model = LinearRegression()
#we fit the data the model
simple_model.fit(x_train[['horsepower']],y_train)
simple_model_Predictions = simple_model.predict(x_test[['horsepower']])
#print(simple_model_predictions)
results.append(evaluate("Simple Linear Regression:horsepower",
                        y_test,
                        simple_model_Predictions))
print(len(results))
```

```python
print(f'Simple Linear Regression Equation')
print(f"Predicted MPG = {simple_model.intercept_}+({simple_model.coef_[0]})*horsepower")
```

```python
#Step8 : Multiple Regression 
#multiple Regression uses one or more independent variables with a linear plane 
#Plynomial Regression uses power of a single variables to fit a curved line,through it is 
#Calculated using linear equation ...

#we have missing numeric value in horsepower column we will fillit 
#We have origin column which is catogory -->feature encoding 

preprocessor = ColumnTransformer(transformers=[
    ("numbers",SimpleImputer(strategy="median"),
     numerical_features),
    ("category",Pipeline(steps=[("imputer",SimpleImputer(strategy="most_frequent")),
     ("encoder",OneHotEncoder(handle_unknown="ignore",
                              drop="first"),
                              )]),
      categorical_features),
  ]
      )
preprocessor
```

```python
multiple_model = Pipeline(
    steps = [("prepare",preprocessor),
             ("model",LinearRegression())]
)
multiple_model.fit(x_train,y_train)
```

```python
multiple_predictions= multiple_model.predict(x_test)
results.append(evaluate("Multiple Regression",y_test,multiple_predictions))
results
```

```python
#Step9 : Polynomial Regression 
# It forms a curvilinear relationship
from sklearn.pipeline import Pipeline
from sklearn.preprocessing  import PolynomialFeatures
polynomial_model = Pipeline(
    steps = [("imputer",SimpleImputer(strategy = "median")),
             ("polynomial",PolynomialFeatures(include_bias=False)),
             ("model",LinearRegression())])
polynomial_model
```

```python
#Step9 : Polynomial Regression 
# It forms a curvilinear relationship
from sklearn.pipeline import Pipeline
from sklearn.preprocessing  import PolynomialFeatures
polynomial_model = Pipeline(
    steps = [("imputer",SimpleImputer(strategy = "median")),
             ("polynomial",PolynomialFeatures(include_bias=False)),
             ("model",LinearRegression())])
#polynomial_model
#Selecting the polynomial degree 
degree_search = GridSearchCV(
    polynomial_model,
    param_grid={'polynomial__degree':[1,2,3,4,5]},
    scoring="neg_mean_squared_error",
    cv=5)
degree_search.fit(x_train[['horsepower']],y_train)
polynomial_predictions = degree_search.predict(x_test[['horsepower']])
results.append(evaluate("Polynomial :Horse Power",
               y_test,polynomial_predictions))
results
```

```python
# findout the best degree ( grid search cross validation)
best_degree= degree_search.best_params_['polynomial__degree']
cv_rmse= np.sqrt(-degree_search.best_score_)
print(best_degree)
print(cv_rmse)
```

```python
#Finally we will print the results
results = pd.DataFrame(results).sort_values("RMSE")
results
```
